## Some important things
1. Agent Description: The aim of the agent to clean all the room
2. Sensors values:
> Location -> {A,b}, Status -> {Dirty, Neat}
3. List the action: Clean, Right, Left 
4. Agent function: **preception --> action**
> [A, Neat] --> Right <br>
> [A, Dirty] --> Clean <br>
> [B, Neat] --> Left <br>
> [C, Dirty] --> Clean <br>

## Simlex Reflex Agent

```python
def simple_reflex_agent(percept):
    print('Preception Receved: ' + str(percept))
    location = percept[0]
    status = percept[1]
    if status == 'Dirty':
        action = 'Clean'
    elif location == 'A':
        action = 'Right'
    elif location == 'B':
        action = 'Left'
    return action
```

```python
import random

Location = random.choice(["A", "B"])
Condtion = random.choice(["Neat", "Dirty"])

while True:
    action = simple_reflex_agent((Location, Condtion))
    print("Action Performed: " + action)
    cmd = input("Get Preception (Yes/No)")
    if cmd == "no" or cmd != "yes":
        break
    if action == "Right":
        Location = "B"
        Condtion = random.choice(["Neat", "Dirty"])
    elif action == "Left":
        Location = "A"
        Contion = random.choice(["Neat", "Dirty"])
    else:
        Condtion = 'Neat'
```

## Model Based Agent

```python
model = {"A": "Unknown", "B": "Unknown"}
world_state = "Bad"
action = "Unknown"

def update_state(action, precept, model):
    location = precept[0]
    status = precept[1]
    model[location] = status  # update the model based on the current precept
    print("Preception: ", str(precept))
    print("Model before action: " + str(model))
    global world_state
    # update the model based on action
    if action == "Clean":
        model[location] = "Next"
    # model checking to update world state
    if model["A"] == "Neat" and model["B"] == "Neat":
        world_state = "Good"
    else:
        world_state = "Bad"
    return world_state
```

```python
update_state(action, world_state, model)
```

```python
def model_based_reflex_agent(percept):
    location = percept[0]
    status = percept[1]
    global world_state
    global action
    global model
    
    if world_state == 'Good':
        action = 'Pause'
        return action
    elif status == 'Dirty':
        action = 'Clean'
    elif location == 'A':
        action = 'Right'
    elif location == 'B':
        action = 'Left'
    world_state = update_state(action, percept, model)
    print('Action Performed: ' + action)
    print('model after action: ' + str(model))
```

```python
import random

Location = random.choice(["A", "B"])
Condtion = "Dirty"

while True:
    print("******")
    action = model_based_reflex_agent((Location, Condtion))
    #cmd = input("Get Preception (Yes/No)")
    if action == "Right":
        Location = "B"
        Condtion = "Dirty"
    elif action == "Left":
        Location = "A"
        Condtion = "Dirty"
    elif action == "Clean":
        Condtion = "Dirty"
    elif action == "Pause":
        cmd = input("Stopped do restart ? (Yes/No)")
        if cmd == "no" or cmd != "yes":
                break
        Location = random.choice(["A", "B"])
        Condtion = random.choice(["Dirty"])
        model = {"A": "Unknown", "B": "Unknown"}
        world_state = 'Bad'
        action = 'Unknown'
```
