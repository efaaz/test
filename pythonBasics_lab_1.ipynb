## Basic & Advance Arithmetic Operation- NUMBER

```python
# addition, sub, multip, div
a = 10
b = 3.5
sum = a + b
sub = a - b
prod = a * b
div = a / b
print("Sum: ", sum)
print("Subtraction: ", sub)
print("Product: ", prod)
print("Division: ", div)
```

```python
print(type(sum))
print(type(div))
newProd = prod * 2
newProd
```

```python
# mod, power, floor, div
import math as m

c = 2
d = 3
mod = d % c
pow = m.pow(c, d)
pow2 = c**d
fdiv = d // c
print(f"mod: {mod}, power_func:{pow}, power_oper:{pow2}, and floor div: {fdiv}")
```

```python
# abs & round
num1 = -7.3
num2 = 3.6
print("abs: ", abs(num1))
print("round num1: ", round(num1))
print("round num2: ", round(num2))
```

```python
# sqrt, factorial, trigonometric value
sqrt = m.sqrt(16)
fact = m.factorial(5)
sin_value = m.sin(m.radians(30))
print(f"Sqrt: {sqrt}, Factorial: {fact}, Sin value: {sin_value}")
```

## Control Flow (if-else)

```python
i = 2
if i < 5:
    print("okay")
    print("fine")
    if i == 2:
        print("got point")
    else:
        print("no point")
else:
    print("bad")
    print("not okay")
```

```python
while True:
    a = input("Enter your Birth Year")
    if a.isdigit():
        print("Your are " + str(2026 - int(a)) + " years old")
        print(2026 - int(a))
        break
    print("Worng input, please enter a valid year.")
```

```python
x = int(input("Enter an integer: "))
if x < 0:
    x = 0
    print("Negative changed to zero.")
elif x == 0:
    print("Zero")
elif x == 1:
    print("Single")
else:
    print("More")
```

## Function

```python
def operate(a, b):
    print(str(a) + ' + ' + str(b) + ' = ' + str(a+b))
    print(str(a) + ' - ' + str(b) + ' = ' + str(a-b))
    print(str(a) + ' x ' + str(b) + ' = ' + str(a*b))
    print(str(a) + ' / ' + str(b) + ' = ' + str(a/b))
```

```python
operate(2,4)
```

```python
def fib(n:int) -> None:
    '''Print a Fibonacci series up to n.'''
    a, b = 0, 1
    while a < n:
        print(a, end=' ')
        a, b = b, a+b
        
```

```python
fib(100000)
```

```python
def addNumbers(a, b):
    c = a + b
    return c
```
